```python
from open_flamingo import create_model_and_transforms

model, image_processor, tokenizer = create_model_and_transforms(
    clip_vision_encoder_path="ViT-L-14",
    clip_vision_encoder_pretrained="openai",
    lang_encoder_path="anas-awadalla/mpt-1b-redpajama-200b",
    tokenizer_path="anas-awadalla/mpt-1b-redpajama-200b",
    cross_attn_every_n_layers=1,
)
# grab model checkpoint from huggingface hub-3b
from huggingface_hub import hf_hub_download
import torch
checkpoint_path = hf_hub_download("openflamingo/OpenFlamingo-3B-vitl-mpt1b", "checkpoint.pt")
model.load_state_dict(torch.load(checkpoint_path), strict=False)
```

```python
from PIL import Image
import requests
import torch
if torch.cuda.is_available():
    print("CUDA is available! Using GPU for calculations.")
    device = torch.device("cuda")
else:
    print("CUDA is not available. Using CPU for calculations.")
    device = torch.device("cpu")
model.to(device)
```

```python
# Dataset directory
DATASET_DIR = '/Users/mohapi.dimakatso/Desktop/Multi_Model/test_data_v2'
CACHE_FILE = './image_descriptions.pkl'

def generate_image_descriptions():
    """Generate textual descriptions for all dataset images and store them."""
    image_description_cache = []
    image_paths = [
        os.path.join(root, file)
        for root, _, files in os.walk(DATASET_DIR)
        for file in files if file.lower().endswith(('.jpg', '.jpeg', '.png'))
    ]
    print(f'Found {len(image_paths)} images. Generating descriptions...')
    for image_path in image_paths:
        try:
            image = Image.open(image_path).convert('RGB')
            vision_x = [image_processor(image).unsqueeze(0)]
            vision_x = torch.cat(vision_x, dim=0)
            vision_x = vision_x.unsqueeze(1).unsqueeze(0).to(device)
            tokenizer.padding_side = 'left'
            lang_x = tokenizer(['<image>This image is'], return_tensors='pt').to(device)
            with torch.no_grad():
                generated_text = model.generate(
                    vision_x=vision_x,
                    lang_x=lang_x['input_ids'],
                    attention_mask=lang_x['attention_mask'],
                    max_new_tokens=50,
                    num_beams=3,
                )
            description = tokenizer.decode(generated_text[0]).strip()
            print(f'Generated description: {description}')
            image_description_cache.append((image_path, description))
        except Exception as e:
            print(f'Error processing image {image_path}: {e}')
    with open(CACHE_FILE, 'wb') as cache_file:
        pickle.dump(image_description_cache, cache_file)
    print(f'Cached {len(image_description_cache)} image descriptions.')

# Run only once unless dataset updates
if not os.path.exists(CACHE_FILE):
    generate_image_descriptions()
```

```python
def search_and_display(user_text):
    """Search for a similar image and display the result."""
    matching_image = find_best_match(user_text)
    if matching_image:
        print(f'Displaying the most similar image for {user_text}: {matching_image}')
        matched_image = Image.open(matching_image)
        plt.figure(figsize=(6, 6))
        plt.imshow(matched_image)
        plt.title(f'Match for {user_text}')
        plt.axis('off')
        plt.show()

# Get user input and run search
user_text = input('Enter a description (e.g., a cat): ')
search_and_display(user_text)
```
